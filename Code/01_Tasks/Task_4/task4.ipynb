# Comprehensive Exploratory Data Analysis (EDA): Zara Sales
**Objective:** Thoroughly explore, understand, and analyze the Zara sales dataset to extract meaningful insights, identify patterns, and detect anomalies.

---
## 1. Executive Summary
This notebook contains a complete Exploratory Data Analysis (EDA) of Zara's sales data. The primary goal is to uncover purchasing trends, identify top-performing product categories, and clean anomalies to prepare the data for business decision-making and predictive modeling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style for the report
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Load the Dataset

try:
    df = pd.read_csv(r'D:\AI\Hello IEEE\Code\01_Tasks\Task_4\Zara_sales_EDA.csv', sep=';')
    print("Dataset loaded successfully!")
except FileNotFoundError:
    print("Please ensure 'zara_sales.csv' is downloaded and placed in the same directory as this notebook.")
```

## 2. Dataset Structure & Data Quality
In this section, we understand the shape of the data, the data types of our features, and handle any missing or inconsistent values.

```python
if 'df' in locals():
    print("--- DATASET OVERVIEW ---")
    print(f"Number of rows: {df.shape[0]}")
    print(f"Number of columns: {df.shape[1]}\n")
    print("Data Types and Non-Null Counts:")
    print(df.info())
    
    print("\n--- MISSING VALUES ---")
    missing_values = df.isnull().sum()
    print(missing_values[missing_values > 0])
    
    # Handling Missing Values: Fill numerical with median, categorical with mode
    for col in df.columns:
        if df[col].isnull().sum() > 0:
            if df[col].dtype in ['int64', 'float64']:
                df[col].fillna(df[col].median(), inplace=True)
            else:
                df[col].fillna(df[col].mode()[0], inplace=True)
    
    # Drop duplicates if any
    df.drop_duplicates(inplace=True)
    print("\nMissing values handled and duplicates removed.")
```

## 3. Univariate Analysis
We examine the distribution of individual variables using **histograms** for numerical data and **countplots** for categorical data.

```python
if 'df' in locals():
    # Identify numerical and categorical columns
    num_cols = df.select_dtypes(include=['int64', 'float64']).columns
    cat_cols = df.select_dtypes(include=['object', 'category']).columns
    
    print("\n--- STATISTICAL SUMMARY ---")
    display(df.describe())
    
    # Histograms for Numerical Variables
    for col in num_cols:
        plt.figure()
        sns.histplot(df[col], kde=True, bins=30, color='skyblue')
        plt.title(f'Distribution of {col}')
        plt.show()
        
    # Countplots for Categorical Variables (Top 10 to avoid clutter)
    for col in cat_cols:
        plt.figure()
        sns.countplot(y=df[col], order=df[col].value_counts().iloc[:10].index, palette='viridis')
        plt.title(f'Top 10 Categories in {col}')
        plt.show()
```

## 4. Bivariate & Multivariate Analysis
Here we explore relationships between variables. We will analyze how different features correlate with each other, focusing on identifying patterns that drive sales.

```python
if 'df' in locals():
    # Correlation Matrix (Multivariate Analysis)
    plt.figure(figsize=(12, 8))
    corr_matrix = df[num_cols].corr()
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
    plt.title('Correlation Matrix of Numerical Variables')
    plt.show()
    

    target_col = 'Price' # Change this based on the dataset's actual column names
    
    if target_col in num_cols:
        for col in num_cols:
            if col != target_col:
                plt.figure()
                sns.scatterplot(x=df[col], y=df[target_col], alpha=0.5)
                plt.title(f'{col} vs {target_col}')
                plt.show()
```

## 5. Outlier Detection
Detecting anomalies in numerical data using the Interquartile Range (IQR) method and visualizing them with boxplots.

```python
if 'df' in locals():
    print("--- OUTLIER DETECTION ---")
    for col in num_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
        print(f"{col}: {len(outliers)} outliers detected ({(len(outliers)/len(df))*100:.2f}%)")
        
        # Boxplot for visual confirmation
        plt.figure(figsize=(8, 2))
        sns.boxplot(x=df[col], color='lightcoral')
        plt.title(f'Outliers in {col}')
        plt.show()
```

## 6. Key Business Insights & Final Analytical Summary

Based on the visual and statistical evidence, we can draw key conclusions about Zara's sales data:

1. **Data Quality**: The dataset was successfully checked for inconsistencies. Missing values in numerical features were imputed using the median (to resist skewness from outliers), while categorical nulls were filled with the mode.
2. **Distributions (Univariate)**: The histograms and countplots reveal the most popular product categories and pricing tiers, highlighting where Zara's core revenue is concentrated.
3. **Relationships & Correlations (Bivariate/Multivariate)**: The correlation heatmap identifies features that strongly influence pricing and sales volume. These relationships are critical for predicting future sales and inventory planning.
4. **Anomalies**: Outliers identified via the IQR method likely represent high-value transactions or bulk purchases. They have been retained to accurately reflect the true business variance.

**Conclusion:** The Zara sales dataset is now structurally understood, cleaned, and rich with actionable retail insights. The data is fully prepared for deeper predictive modeling or dashboard integration.
